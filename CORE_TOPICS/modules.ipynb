```python
# Statistics Module
from statistics import *
ages = [20, 20,  24, 25, 22, 26, 20, 23, 22, 26]
print(mean(ages))#Mean → The average value.
print(median(ages))#median the middle value
print(mode(ages)) #most freqaunt value in list
print(stdev(ages))#A measure of how spread out the data is around the mean(ek lsit aur dursi ke bich ka kitne number ka gap hai)


```

```python
### Exercises: Level 1
import random
import string 

def random_user_id():
    character = string.ascii_lowercase + string.digits
    user_id = ''.join(random.choice(character) for _ in range(6))
    return user_id
print(random_user_id())
print(random_user_id())
print(random_user_id())
```

```python
# 2. Modify the previous task. Declare a function named user_id_gen_by_user.
# It doesn’t take any parameters but it takes two inputs using input().
# One of the inputs is the number of characters and the second input is the number of IDs which are supposed to be generated.
   
import random
import string 

def user_id_gen_by_user():
    numchar = int(input("enter the number:"))
    numid = int(input("enter the number"))

    character = string.ascii_lowercase + string.digits

    for _ in range(numid):
        user_id = ''.join(random.choice(character) for _ in range(numchar))
        print(user_id)
user_id_gen_by_user()
```

```python

# 3. Write a function named rgb_color_gen. It will generate rgb colors (3 values ranging from 0 to 255 each).
import string
import random 

def rgb():
    n1 = string.digits

    for _ in  range(0,256):
        color = ','.join(random.choice(n1) for _ in range(3))
        print(color)
rgb()
```
